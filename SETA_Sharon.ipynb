```python
class Product:
    def __init__(self, product_id, name, price, stock):
        self.product_id = product_id
        self.name = name
        self.price = price
        self.stock = stock

class ShoppingCart:
    def __init__(self, products):
        self.products = products
        self.cart = {}

    def add_product(self, product_id, quantity):
        if product_id not in self.products:
            print("Product does not exist")
            return

        if quantity <= 0:
            print("Invalid quantity")
            return

        product = self.products[product_id]


        current_quantity = self.cart.get(product_id, 0)

        if current_quantity + quantity > product.stock:
            print("Insufficient stock")
            return

        self.cart[product_id] = current_quantity + quantity

    def remove_product(self, product_id, quantity):
        if product_id not in self.cart:
            print("Product not in cart")
            return

        if quantity <= 0:
            print("Invalid quantity")
            return

        if quantity >= self.cart[product_id]:
            del self.cart[product_id]
        else:
            self.cart[product_id] -= quantity

    def calculate_total(self):
        total = 0

        for product_id, quantity in self.cart.items():
            product = self.products[product_id]
            total += product.price * quantity

        return total

    def apply_discount(self):
        total = self.calculate_total()

        if total >= 20000:
            discount_percent = 15
        elif total >= 10000:
            discount_percent = 10
        elif total >= 5000:
            discount_percent = 5
        else:
            discount_percent = 0

        discount_amount = total * discount_percent // 100
        final_amount = total - discount_amount

        return discount_percent, discount_amount, final_amount

    def display_cart(self):
        if not self.cart:
            print("Cart is empty")
            return

        for product_id, quantity in self.cart.items():
            product = self.products[product_id]
            subtotal = product.price * quantity

            print(
                product.name,
                "- Quantity:", quantity,
                "- Subtotal:", subtotal
            )

        total = self.calculate_total()
        discount_percent, discount_amount, final_amount = self.apply_discount()

        print("Total Amount:", total)
        print("Discount:", str(discount_percent) + "%")
        print("Discount Amount:", discount_amount)
        print("Final Amount:", final_amount)



n = int(input("Enter the number of Products"))
products = {}
for _ in range(n):
    product_id, name, price, stock = input().split()

    product_id = int(product_id)
    price = int(price)
    stock = int(stock)

    products[product_id] = Product(
        product_id, name, price, stock
    )


cart = ShoppingCart(products)

operations = int(input("Enter  the number of operations"))

for _ in range(operations):
    operation = input().split()

    if operation[0] == "ADD":
        product_id = int(operation[1])
        quantity = int(operation[2])

        cart.add_product(product_id, quantity)

    elif operation[0] == "REMOVE":
        product_id = int(operation[1])
        quantity = int(operation[2])

        cart.remove_product(product_id, quantity)

    elif operation[0] == "TOTAL":
        print("Total:", cart.calculate_total())

    elif operation[0] == "DISPLAY":
        cart.display_cart()
```
